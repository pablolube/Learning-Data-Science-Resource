# 📚 Pilas y Colas – Apunte Teórico

## 🧱 PILAS (Stacks)

### 📌 Definición
Una **pila** es una **estructura de datos lineal** en la que los elementos se agregan y se eliminan **solo por un extremo**, llamado **tope**.

Funciona según el principio:

> **LIFO** → *Last In, First Out*  
> El último en entrar es el primero en salir.

---

### 🔁 Operaciones básicas

| Operación | Descripción |
|----------|-------------|
| **push** | Agrega un elemento al tope de la pila |
| **pop** | Elimina el elemento del tope |
| **peek / top** | Devuelve el elemento del tope sin eliminarlo |
| **isEmpty** | Verifica si la pila está vacía |
| **size** | Devuelve la cantidad de elementos |

---

### 🧠 Características
- Acceso restringido (solo al tope)
- No permite acceso directo a elementos intermedios
- Implementable con arrays o listas enlazadas
- Estructura simple y eficiente

---

### 🧩 Casos de uso
- Deshacer / Rehacer
- Pila de llamadas de funciones
- Evaluación de expresiones matemáticas
- Navegación (volver atrás)
- Validación de paréntesis

---

### ⚠️ Errores comunes
- **Stack Overflow**: intentar agregar en una pila llena
- **Stack Underflow**: intentar quitar de una pila vacía

---

## 🚶 COLAS (Queues)

### 📌 Definición
Una **cola** es una **estructura de datos lineal** en la que los elementos se agregan por un extremo (**final**) y se eliminan por el otro (**frente**).

Funciona según el principio:

> **FIFO** → *First In, First Out*  
> El primero en entrar es el primero en salir.

---

### 🔁 Operaciones básicas

| Operación | Descripción |
|----------|-------------|
| **enqueue** | Agrega un elemento al final de la cola |
| **dequeue** | Elimina el elemento del frente |
| **front / peek** | Devuelve el primer elemento sin eliminarlo |
| **isEmpty** | Verifica si la cola está vacía |
| **size** | Devuelve la cantidad de elementos |

---

### 🧠 Características
- Inserción y eliminación por extremos distintos
- Mantiene el orden de llegada
- Ideal para procesamiento secuencial
- Implementable con arrays o listas enlazadas

---

---

### 🧩 Casos de uso
- Sistemas de turnos
- Impresoras
- Procesos del sistema operativo
- Manejo de eventos
- Buffers de datos

---

### ⚠️ Errores comunes
- **Queue Overflow**: intentar agregar en una cola llena
- **Queue Underflow**: intentar quitar de una cola vacía

---

## 🔄 Comparación Pilas vs Colas

| Característica | Pila | Cola |
|---------------|------|------|
| Principio | LIFO | FIFO |
| Inserción | Tope | Final |
| Eliminación | Tope | Frente |
| Uso típico | Retroceso | Espera |

---

## 🧠 Resumen
- **Pila**: se utiliza cuando importa el último elemento agregado
- **Cola**: se utiliza cuando importa el orden de llegada
- Ambas son estructuras fundamentales en programación y algoritmos
