# Diferencia entre `sort()` y `sorted()` en Python

## 🔹 `list.sort()`

- Es un **método de las listas**
- **Modifica la lista original** (*in-place*)
- **Devuelve `None`**
- Solo funciona con **listas**

```python
numeros = [3, 1, 4, 2]
numeros.sort()
print(numeros)


| Característica            | `sort()`        | `sorted()`    |
| ------------------------- | --------------- | ------------- |
| Tipo                      | Método de lista | Función       |
| Modifica el original      | ✅ Sí            | ❌ No          |
| Valor de retorno          | ❌ `None`        | ✅ Nueva lista |
| Acepta cualquier iterable | ❌ No            | ✅ Sí          |
