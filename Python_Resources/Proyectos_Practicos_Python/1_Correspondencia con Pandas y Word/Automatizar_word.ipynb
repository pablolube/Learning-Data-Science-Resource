```python
import pandas as pd
from datetime import datetime
from docxtpl import DocxTemplate
 
```

```python
doc=DocxTemplate("plantilla.docx")
```

```python

nombre = "Pablo"
apellido="Luberriaga"
correo= "Pablolube@gmail.com"
telefono = "2214882037"
fecha=datetime.today().strftime("%d/%m/%Y")
```

```python
constantes={"nombre":nombre,"apellido":apellido,"telefono":telefono,"correo":correo,"fecha":fecha}
print(constantes)

```

```python
df=pd.read_excel("notas.xlsx")
df
```

```python
for indice,fila in df.iterrows():
    contenido={"nombre_alumno":fila["alumno"],
               "mat":fila["mat"],"qui":fila["qui"],"fis":fila["fis"]}
    contenido.update(constantes)

    doc.render(contenido)
    doc.save(f"notas_de_{fila["alumno"]}.docx")
    
```
